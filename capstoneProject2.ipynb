```python
from pandas_datareader import data,wb
import pandas as pd
import seaborn as sns
import yfinance as yf
import numpy as np
import datetime
%matplotlib inline
```

```python
start= datetime.datetime(2006,1,1)
end = datetime.datetime(2016,1,1)
```

```python
BAC = yf.download('BAC', start='2006-01-01', end='2023-01-01') # here are some symbol of which we have created this dfs 
C= yf.download('C', start='2006-01-01', end='2023-01-01')
GS = yf.download('GS', start='2006-01-01', end='2023-01-01')
MS = yf.download('MS', start='2006-01-01', end='2023-01-01')
JPM= yf.download('JPM', start='2006-01-01', end='2023-01-01')
WFC = yf.download('WFC', start='2006-01-01', end='2023-01-01')
```

```python
tickers=['BAC','C','GS','MS','JPM','WFC']
```

```python
banks_stocks = pd.concat([BAC,C,GS,MS,JPM,WFC],axis=1,keys=tickers) #here we created only on df for all 
```

```python
banks_stocks.head()
```

```python

banks_stocks.columns.names=['Bank_Ticker','Stock_Info','Tickers']
# assigned then name to columns 
```

```python
banks_stocks.head()
```

```python
banks_stocks.columns
banks_stocks.columns.nlevels
```

```python
for ticks in tickers:
    print(ticks,banks_stocks[ticks]['Close'].max()) #this code is to check high close for each symbol or ticker we can say
```

```python
returns = pd.DataFrame()
```

```python
for ticks in tickers: #filling data in returns df only for close price
    returns[ticks+' Return']=banks_stocks[ticks]['Close'].pct_change()
```

```python
returns.head()
```

```python
sns.pairplot(returns[1:]) #we get crash at MS stock some where
```
